```python
#[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
# formatting this list of dictionary to list:
#['Action', 'Adventue','Fantasy','SciFi']
```

```python
# def convert(obj):
#     list = []
#     for i in obj:
#         list.append(i['name'])
#         return list

# error will come because we get the object it will be in the form of string we have to convert it to list using literal_eval
import ast
def convert(obj):
    list =[]
    for i in ast.literal_eval(obj):
        list.append(i['name'])
    return list
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies.head(1)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head(1)
```

```python
# cast , we want top 3 actors so we take first 3 dictionaries

def getThreeCast(obj):
    counter = 0
    list =[]
    for i in ast.literal_eval(obj):
        if counter != 3:
            list.append(i['name'])
            counter+=1
        else:
            break
    return list
```

```python
movies['cast'] = movies['cast'].apply(getThreeCast)
```

```python
def getDirector(obj):
    list = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            list.append(i['name'])
            break
    return list

movies['crew'] = movies['crew'].apply(getDirector)
movies['crew'].values
```

```python
movies.head(1)
```

```python
# converting overview string to list
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head(1)
# now concatinating these lists and converting them to string
```

```python
# apply transformation
# remove space between words and names 
# like we have Sam Worthington and Sam Mendence we will have 2 tags for Sam and model will be confused but by removing space we tackle this problem


movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head(1)
#space removed 
```

```python
# concatinatings to one tags column

movies['tags'] = movies['overview']+movies['genres']+movies['keywords']+movies['crew']
movies.iloc[0].tags
```

```python
new_df = movies[['movie_id','title','tags']]
new_df
```

```python
# converting tags list to string
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df.iloc[0].tags
```

```python
# converting to lower case
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
new_df.iloc[0].tags
```

```python
new_df['tags'][1]
```

```python
# text vectorization
# get similar 5 movies
# convert tags to vectors
# by doing this every movie will become a vector
# now we have to see the closest vectors to get similar movies
# different techniques:
# bag of words (we are using)
# word2vec
# in bag of words we combine all the words(tags)
```

```python
#   get the most common words from all tags
# get more data from minimum number of words
# as no. of words increase data dimensionality increase
# we dont considered stop words (e.g are,and, of,is,to,the) 
# this work is done by countvectorizer
```

```python
# now we got problem like we have same type of words also
# like action, actions, activity, activities
# so apply stemming  technique 
# what it it do
# [loved, lover, love] to [love, love,love]
```

```python
import nltk   #(pip install nltk)
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)   
    # return data in string
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
# countVectorizer returns sciPy sparse matrix we have to convert it into numpy array
vectors = cv.fit_transform(new_df['tags']).toarray()
vectors.shape
# we have 4806 movies and each movie have 5000 words
# now every movie is in vector form
```

```python
# we can see 5000 most frequent words
cv.get_features_names()
```

```python
# 4806 vectors and per vector 5000 words
# every movie is vector
# now we have to calculate distance from every movie to other
# euclidean difference calculates difference betweern tips of two vectors but it has problem while working in large dataset
# we will calculate cosine differnce (angle (theta) between them)
# angle 0 same movie, angle 180 opposite movies
```

```python
from sklearn.metrics.pairwise import cosine_similarity
# tells value between 0 and 1 , closer to 1 more similar
```

```python
similarity = cosine_similarity(vectors)
similarity[0]  
# matrix tells similarity score (differnces) of first movie with every other movie
```

```python
sorted(list(enumerate(similarity[0])),reverse=True, key= lambda x:x[1])[1:6]

# key= lambda x:x[1] telling that we want sorting based on second thing (similarity score) not first thing (index)
# [1:6] get first five
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]     # get the movies index from all movies
    #print(new_df[new_df['title'] == movie].index) :-  Index([119], dtype='int64')
    distances = similarity[movie_index]      # get distances of this index movie from all other movies
    movies_list = sorted(list(enumerate(distances)),reverse=True, key= lambda x:x[1])[1:6]      # sort in descending acc to similarity and get first 5 movies

    # movies list have list indexes of movie and similarity 
    #e.g (65, np.float64(0.3885143449429058))
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Aliens')
```
